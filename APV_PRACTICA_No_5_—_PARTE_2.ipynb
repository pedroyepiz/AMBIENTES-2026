### Procedimiento básico para guardar trabajo con GitHub Desktop

Cuando una actividad indique **Commit + Push**, realizar:

1. Guardar todos los archivos en Visual Studio:

   **File → Save All**

2. Abrir GitHub Desktop.

3. Revisar arriba de la ventana:

   **Current Branch**

   y confirmar que sea la rama correspondiente al rol.

4. En el panel izquierdo revisar los archivos de **Changes**.

5. Confirmar que los archivos modificados corresponden a la actividad realizada.

6. En **Summary (required)** escribir un mensaje descriptivo.

7. Presionar:

   **Commit to frontend**, **Commit to backend** o **Commit to dba**, según corresponda.

8. Después del Commit, presionar:

   **Push origin**

9. Esperar a que termine.

10. Avisar al equipo cuando el cambio esté disponible en GitHub.


```text
1. AVISAR
2. ACORDAR
3. MODIFICAR
4. COMMIT
5. PUSH
6. INFORMAR QUE EL CAMBIO ESTÁ DISPONIBLE
```

> **No cambiar sin avisar**
>
> Ejemplo: si Frontend usa Telefono, Backend no debe cambiarlo a Celular y DBA no debe crear numero_contacto. Los tres deben conservar el contrato común.

## C-03. Crear la solución sin duplicar carpetas

En esta etapa trabajará el **Backend**, mientras Frontend y DBA observan.

> **Punto crítico**
>
> La plantilla **Blank Solution** normalmente crea una carpeta con el mismo nombre de la solución dentro de la ruta indicada en **Location**.
>
> Como el repositorio ya se llama `Veterinaria`, en **Location NO se debe seleccionar la carpeta `Veterinaria`**.  
> Se debe seleccionar **la carpeta que contiene al repositorio**.


### Ejemplo correcto

Si el repositorio está en:

```text
C:\Users\Pedro\Documents\GitHub\Veterinaria
```

entonces:

```text
Solution name: Veterinaria
Location:      C:\Users\Pedro\Documents\GitHub
```

El resultado esperado es:

```text
C:\Users\Pedro\Documents\GitHub\Veterinaria\Veterinaria.sln
```


### Ejemplo incorrecto

Si se escribe:

```text
Location:
C:\Users\Pedro\Documents\GitHub\Veterinaria
```

Visual Studio puede crear:

```text
C:\Users\Pedro\Documents\GitHub\Veterinaria\Veterinaria\Veterinaria.sln
```

y quedaría una carpeta duplicada.


### Paso a paso

1. Abrir **Visual Studio 2026**.

2. Seleccionar:

   **Create a new project / Crear un nuevo proyecto**

3. Buscar:

   `Blank Solution`

4. Seleccionar:

   **Blank Solution / Solución en blanco**

5. Presionar **Next**.

6. En **Solution name** escribir:

   `Veterinaria`

7. En **Location**, presionar `...`.

8. Localizar la carpeta del repositorio.

9. **Subir un nivel** y seleccionar la carpeta que contiene al repositorio.

   Ejemplo:

   seleccionar:

   `C:\Users\Pedro\Documents\GitHub`

   y NO:

   `C:\Users\Pedro\Documents\GitHub\Veterinaria`

10. Presionar **Create**.

11. Abrir:

    **View → Solution Explorer**

12. Guardar:

    **File → Save All**

13. Regresar a GitHub Desktop y seleccionar:

    **Repository → Show in Explorer**

14. Confirmar que exista:

```text
Veterinaria/
├── .gitignore
├── README.md
└── Veterinaria.sln
```

15. Si existe una segunda carpeta:

```text
Veterinaria/Veterinaria/
```

detenerse y corregir la ubicación antes de agregar proyectos.


### Si ya se creó una carpeta duplicada

Si por error quedó algo como:

```text
Veterinaria/
└── Veterinaria/
    └── Veterinaria.sln
```

y **todavía no se han creado los cuatro proyectos ni se ha escrito código**, realizar:

8. Volver a realizar C-03 usando como **Location** la carpeta padre del repositorio.
9. Confirmar que `Veterinaria.sln` quede directamente en la raíz.

> **No arrastrar manualmente proyectos o carpetas si ya existe trabajo importante.**

```csharp
using System.ComponentModel.DataAnnotations;

namespace Veterinaria.Entidades;

public class Mascota
{
    public int IdMascota { get; set; }

    [Range(1, int.MaxValue, ErrorMessage = "Debe seleccionar un propietario.")]
    public int IdCliente { get; set; }

    [Required(ErrorMessage = "El nombre de la mascota es obligatorio.")]
    [StringLength(60)]
    public string Nombre { get; set; } = string.Empty;

    [Required(ErrorMessage = "La especie es obligatoria.")]
    [StringLength(40)]
    public string Especie { get; set; } = string.Empty;

    [StringLength(60)]
    public string? Raza { get; set; }

    [StringLength(15)]
    public string? Sexo { get; set; }

    public DateTime? FechaNacimiento { get; set; }

    [StringLength(50)]
    public string? Color { get; set; }

    [StringLength(200)]
    public string? SenasParticulares { get; set; }
}
```

> **Sobre selección del propietario**
>
> `IdCliente` es `int`, por eso el `<option>` inicial del `<select>` utiliza `value="0"` y no `value=""`.
>
> Así el model binding recibe un entero válido (`0`) y el atributo:
>
> ```csharp
> [Range(1, int.MaxValue,
>     ErrorMessage = "Debe seleccionar un propietario.")]
> ```
>
> puede mostrar el mensaje en español. Si se usa `value=""`, ASP.NET puede generar primero un error de conversión como `The value '' is invalid`.
>
> **Sobre Edad**
>
> No se almacenará Edad en la tabla Mascota. Se guardará FechaNacimiento porque la edad cambia con el tiempo y puede calcularse cuando sea necesaria.

```csharp
using Microsoft.AspNetCore.Mvc;
using Microsoft.AspNetCore.Mvc.RazorPages;
using Veterinaria.Entidades;

namespace Veterinaria.Web.Pages.Clientes;

public class IndexModel : PageModel
{
    [BindProperty(SupportsGet = true)]
    public string? Buscar { get; set; }

    public List<Cliente> Clientes { get; private set; } = new();

    public void OnGet()
    {
        Clientes = new List<Cliente>
        {
            new()
            {
                IdCliente = 1,
                Nombre = "Ana",
                ApellidoPaterno = "López",
                ApellidoMaterno = "García",
                Telefono = "6461234567",
                Correo = "ana@correo.com",
                Direccion = "Ensenada"
            },
            new()
            {
                IdCliente = 2,
                Nombre = "Carlos",
                ApellidoPaterno = "Ramírez",
                ApellidoMaterno = "Torres",
                Telefono = "6467654321",
                Correo = "carlos@correo.com",
                Direccion = "Ensenada"
            }
        };

        if (!string.IsNullOrWhiteSpace(Buscar))
        {
            string texto = Buscar.Trim();

            Clientes = Clientes
                .Where(c =>
                    c.Nombre.Contains(texto, StringComparison.OrdinalIgnoreCase) ||
                    c.ApellidoPaterno.Contains(texto, StringComparison.OrdinalIgnoreCase) ||
                    (c.Correo?.Contains(texto, StringComparison.OrdinalIgnoreCase) ?? false) ||
                    c.Telefono.Contains(texto, StringComparison.OrdinalIgnoreCase))
                .ToList();
        }
    }
}
```

```csharp
using MySql.Data.MySqlClient;

namespace Veterinaria.Datos;

public static class Conexion
{
    private static string? _cadenaConexion;

    public static void Configurar(string cadenaConexion)
    {
        _cadenaConexion = cadenaConexion;
    }

    public static MySqlConnection Crear()
    {
        if (string.IsNullOrWhiteSpace(_cadenaConexion))
        {
            throw new InvalidOperationException(
                "La cadena de conexión de Veterinaria no ha sido configurada.");
        }

        return new MySqlConnection(_cadenaConexion);
    }

    public static int Probar()
    {
        using MySqlConnection cn = Crear();
        cn.Open();

        using MySqlCommand cmd =
            new("SELECT COUNT(*) FROM cliente;", cn);

        return Convert.ToInt32(cmd.ExecuteScalar());
    }
}
```

```csharp
using MySql.Data.MySqlClient;
using Veterinaria.Entidades;

namespace Veterinaria.Datos;

public class ClienteDatos
{
    public List<Cliente> Listar()
    {
        List<Cliente> lista = new();

        using MySqlConnection cn = Conexion.Crear();
        cn.Open();

        const string sql = """
            SELECT id_cliente, nombre, apellido_paterno,
                   apellido_materno, telefono, correo,
                   direccion, fecha_registro
            FROM cliente
            ORDER BY apellido_paterno, nombre;
            """;

        using MySqlCommand cmd = new(sql, cn);
        using MySqlDataReader dr = cmd.ExecuteReader();

        while (dr.Read())
        {
            lista.Add(Mapear(dr));
        }

        return lista;
    }

    public Cliente? BuscarPorId(int idCliente)
    {
        using MySqlConnection cn = Conexion.Crear();
        cn.Open();

        const string sql = """
            SELECT id_cliente, nombre, apellido_paterno,
                   apellido_materno, telefono, correo,
                   direccion, fecha_registro
            FROM cliente
            WHERE id_cliente = @idCliente;
            """;

        using MySqlCommand cmd = new(sql, cn);
        cmd.Parameters.AddWithValue("@idCliente", idCliente);

        using MySqlDataReader dr = cmd.ExecuteReader();

        return dr.Read() ? Mapear(dr) : null;
    }

    public List<Cliente> Buscar(string texto)
    {
        List<Cliente> lista = new();

        using MySqlConnection cn = Conexion.Crear();
        cn.Open();

        const string sql = """
            SELECT id_cliente, nombre, apellido_paterno,
                   apellido_materno, telefono, correo,
                   direccion, fecha_registro
            FROM cliente
            WHERE nombre LIKE @texto
               OR apellido_paterno LIKE @texto
               OR apellido_materno LIKE @texto
               OR telefono LIKE @texto
               OR correo LIKE @texto
            ORDER BY apellido_paterno, nombre;
            """;

        using MySqlCommand cmd = new(sql, cn);
        cmd.Parameters.AddWithValue("@texto", $"%{texto}%");

        using MySqlDataReader dr = cmd.ExecuteReader();

        while (dr.Read())
        {
            lista.Add(Mapear(dr));
        }

        return lista;
    }

    public int Agregar(Cliente cliente)
    {
        using MySqlConnection cn = Conexion.Crear();
        cn.Open();

        const string sql = """
            INSERT INTO cliente
            (nombre, apellido_paterno, apellido_materno,
             telefono, correo, direccion)
            VALUES
            (@nombre, @apellidoPaterno, @apellidoMaterno,
             @telefono, @correo, @direccion);
            """;

        using MySqlCommand cmd = new(sql, cn);

        AgregarParametros(cmd, cliente);
        cmd.ExecuteNonQuery();

        return Convert.ToInt32(cmd.LastInsertedId);
    }

    public bool Modificar(Cliente cliente)
    {
        using MySqlConnection cn = Conexion.Crear();
        cn.Open();

        const string sql = """
            UPDATE cliente
            SET nombre = @nombre,
                apellido_paterno = @apellidoPaterno,
                apellido_materno = @apellidoMaterno,
                telefono = @telefono,
                correo = @correo,
                direccion = @direccion
            WHERE id_cliente = @idCliente;
            """;

        using MySqlCommand cmd = new(sql, cn);

        AgregarParametros(cmd, cliente);
        cmd.Parameters.AddWithValue("@idCliente", cliente.IdCliente);

        return cmd.ExecuteNonQuery() > 0;
    }

    private static void AgregarParametros(
        MySqlCommand cmd,
        Cliente cliente)
    {
        cmd.Parameters.AddWithValue("@nombre", cliente.Nombre);
        cmd.Parameters.AddWithValue(
            "@apellidoPaterno", cliente.ApellidoPaterno);

        cmd.Parameters.AddWithValue(
            "@apellidoMaterno",
            (object?)cliente.ApellidoMaterno ?? DBNull.Value);

        cmd.Parameters.AddWithValue("@telefono", cliente.Telefono);

        cmd.Parameters.AddWithValue(
            "@correo",
            (object?)cliente.Correo ?? DBNull.Value);

        cmd.Parameters.AddWithValue(
            "@direccion",
            (object?)cliente.Direccion ?? DBNull.Value);
    }

    private static Cliente Mapear(MySqlDataReader dr)
    {
        return new Cliente
        {
            IdCliente = Convert.ToInt32(dr["id_cliente"]),
            Nombre = Convert.ToString(dr["nombre"]) ?? "",
            ApellidoPaterno =
                Convert.ToString(dr["apellido_paterno"]) ?? "",
            ApellidoMaterno =
                dr["apellido_materno"] == DBNull.Value
                    ? null
                    : Convert.ToString(dr["apellido_materno"]),
            Telefono = Convert.ToString(dr["telefono"]) ?? "",
            Correo =
                dr["correo"] == DBNull.Value
                    ? null
                    : Convert.ToString(dr["correo"]),
            Direccion =
                dr["direccion"] == DBNull.Value
                    ? null
                    : Convert.ToString(dr["direccion"]),
            FechaRegistro =
                Convert.ToDateTime(dr["fecha_registro"])
        };
    }
}
```

```csharp
using MySql.Data.MySqlClient;
using Veterinaria.Entidades;

namespace Veterinaria.Datos;

public class MascotaDatos
{
    public List<Mascota> Listar()
    {
        List<Mascota> lista = new();

        using MySqlConnection cn = Conexion.Crear();
        cn.Open();

        const string sql = """
            SELECT id_mascota, id_cliente, nombre, especie,
                   raza, sexo, fecha_nacimiento, color,
                   senas_particulares
            FROM mascota
            ORDER BY nombre;
            """;

        using MySqlCommand cmd = new(sql, cn);
        using MySqlDataReader dr = cmd.ExecuteReader();

        while (dr.Read())
            lista.Add(Mapear(dr));

        return lista;
    }

    public List<Mascota> ListarPorCliente(int idCliente)
    {
        List<Mascota> lista = new();

        using MySqlConnection cn = Conexion.Crear();
        cn.Open();

        const string sql = """
            SELECT id_mascota, id_cliente, nombre, especie,
                   raza, sexo, fecha_nacimiento, color,
                   senas_particulares
            FROM mascota
            WHERE id_cliente = @idCliente
            ORDER BY nombre;
            """;

        using MySqlCommand cmd = new(sql, cn);
        cmd.Parameters.AddWithValue("@idCliente", idCliente);

        using MySqlDataReader dr = cmd.ExecuteReader();

        while (dr.Read())
            lista.Add(Mapear(dr));

        return lista;
    }

    public Mascota? BuscarPorId(int idMascota)
    {
        using MySqlConnection cn = Conexion.Crear();
        cn.Open();

        const string sql = """
            SELECT id_mascota, id_cliente, nombre, especie,
                   raza, sexo, fecha_nacimiento, color,
                   senas_particulares
            FROM mascota
            WHERE id_mascota = @idMascota;
            """;

        using MySqlCommand cmd = new(sql, cn);
        cmd.Parameters.AddWithValue("@idMascota", idMascota);

        using MySqlDataReader dr = cmd.ExecuteReader();

        return dr.Read() ? Mapear(dr) : null;
    }

    public int Agregar(Mascota mascota)
    {
        using MySqlConnection cn = Conexion.Crear();
        cn.Open();

        const string sql = """
            INSERT INTO mascota
            (id_cliente, nombre, especie, raza, sexo,
             fecha_nacimiento, color, senas_particulares)
            VALUES
            (@idCliente, @nombre, @especie, @raza, @sexo,
             @fechaNacimiento, @color, @senasParticulares);
            """;

        using MySqlCommand cmd = new(sql, cn);
        AgregarParametros(cmd, mascota);

        cmd.ExecuteNonQuery();
        return Convert.ToInt32(cmd.LastInsertedId);
    }

    public bool Modificar(Mascota mascota)
    {
        using MySqlConnection cn = Conexion.Crear();
        cn.Open();

        const string sql = """
            UPDATE mascota
            SET id_cliente = @idCliente,
                nombre = @nombre,
                especie = @especie,
                raza = @raza,
                sexo = @sexo,
                fecha_nacimiento = @fechaNacimiento,
                color = @color,
                senas_particulares = @senasParticulares
            WHERE id_mascota = @idMascota;
            """;

        using MySqlCommand cmd = new(sql, cn);
        AgregarParametros(cmd, mascota);
        cmd.Parameters.AddWithValue("@idMascota", mascota.IdMascota);

        return cmd.ExecuteNonQuery() > 0;
    }

    private static void AgregarParametros(
        MySqlCommand cmd,
        Mascota mascota)
    {
        cmd.Parameters.AddWithValue("@idCliente", mascota.IdCliente);
        cmd.Parameters.AddWithValue("@nombre", mascota.Nombre);
        cmd.Parameters.AddWithValue("@especie", mascota.Especie);

        cmd.Parameters.AddWithValue(
            "@raza", (object?)mascota.Raza ?? DBNull.Value);

        cmd.Parameters.AddWithValue(
            "@sexo", (object?)mascota.Sexo ?? DBNull.Value);

        cmd.Parameters.AddWithValue(
            "@fechaNacimiento",
            (object?)mascota.FechaNacimiento ?? DBNull.Value);

        cmd.Parameters.AddWithValue(
            "@color", (object?)mascota.Color ?? DBNull.Value);

        cmd.Parameters.AddWithValue(
            "@senasParticulares",
            (object?)mascota.SenasParticulares ?? DBNull.Value);
    }

    private static Mascota Mapear(MySqlDataReader dr)
    {
        return new Mascota
        {
            IdMascota = Convert.ToInt32(dr["id_mascota"]),
            IdCliente = Convert.ToInt32(dr["id_cliente"]),
            Nombre = Convert.ToString(dr["nombre"]) ?? "",
            Especie = Convert.ToString(dr["especie"]) ?? "",
            Raza =
                dr["raza"] == DBNull.Value
                    ? null
                    : Convert.ToString(dr["raza"]),
            Sexo =
                dr["sexo"] == DBNull.Value
                    ? null
                    : Convert.ToString(dr["sexo"]),
            FechaNacimiento =
                dr["fecha_nacimiento"] == DBNull.Value
                    ? null
                    : Convert.ToDateTime(dr["fecha_nacimiento"]),
            Color =
                dr["color"] == DBNull.Value
                    ? null
                    : Convert.ToString(dr["color"]),
            SenasParticulares =
                dr["senas_particulares"] == DBNull.Value
                    ? null
                    : Convert.ToString(dr["senas_particulares"])
        };
    }
}
```

> **Nota sobre `FechaNacimiento`: C# y MySQL**
>
> En `Mascota.cs` se utiliza:
>
> ```csharp
> public DateTime? FechaNacimiento { get; set; }
> ```
>
> y en MySQL:
>
> ```sql
> fecha_nacimiento DATE
> ```
>
> Esta combinación es correcta para esta práctica. El tipo `DATE` almacena únicamente año, mes y día; `MySql.Data` puede leerlo como `DateTime`. Como la propiedad es nullable (`DateTime?`), el código de `MascotaDatos.cs` primero comprueba `DBNull.Value` y sólo después utiliza `Convert.ToDateTime(...)`.
>
> El control Razor:
>
> ```html
> <input asp-for="Mascota.FechaNacimiento"
>        type="date"
>        class="form-control" />
> ```
>
> también es compatible con `DateTime?`.
>
> Por ello **no se debe cambiar `FechaNacimiento` a `string`**. Si el campo está vacío se manejará como `null`; si contiene una fecha, la parte de hora no se utiliza en la base de datos.

```csharp
using Veterinaria.Datos;
using Veterinaria.Entidades;

namespace Veterinaria.Negocios;

public class ClienteNegocio
{
    private readonly ClienteDatos _datos = new();

    public List<Cliente> Listar()
        => _datos.Listar();

    public List<Cliente> Buscar(string texto)
    {
        if (string.IsNullOrWhiteSpace(texto))
            return _datos.Listar();

        return _datos.Buscar(texto.Trim());
    }

    public Cliente? BuscarPorId(int idCliente)
        => _datos.BuscarPorId(idCliente);

    public int Agregar(Cliente cliente)
    {
        Validar(cliente);
        return _datos.Agregar(cliente);
    }

    public bool Modificar(Cliente cliente)
    {
        if (cliente.IdCliente <= 0)
            throw new ArgumentException("Cliente no válido.");

        Validar(cliente);
        return _datos.Modificar(cliente);
    }

    private static void Validar(Cliente cliente)
    {
        if (string.IsNullOrWhiteSpace(cliente.Nombre))
            throw new ArgumentException("El nombre es obligatorio.");

        if (string.IsNullOrWhiteSpace(cliente.ApellidoPaterno))
            throw new ArgumentException(
                "El apellido paterno es obligatorio.");

        if (string.IsNullOrWhiteSpace(cliente.Telefono))
            throw new ArgumentException("El teléfono es obligatorio.");
    }
}
```

```sql
DROP DATABASE IF EXISTS Veterinaria;

CREATE DATABASE Veterinaria
CHARACTER SET utf8mb4
COLLATE utf8mb4_unicode_ci;

USE Veterinaria;

CREATE TABLE rol
(
    id_rol INT AUTO_INCREMENT PRIMARY KEY,
    nombre VARCHAR(30) NOT NULL UNIQUE,
    descripcion VARCHAR(150)
);

CREATE TABLE cliente
(
    id_cliente INT AUTO_INCREMENT PRIMARY KEY,
    nombre VARCHAR(60) NOT NULL,
    apellido_paterno VARCHAR(60) NOT NULL,
    apellido_materno VARCHAR(60),
    telefono VARCHAR(20) NOT NULL,
    correo VARCHAR(120),
    direccion VARCHAR(200),
    fecha_registro DATETIME NOT NULL DEFAULT CURRENT_TIMESTAMP
);

CREATE TABLE veterinario
(
    id_veterinario INT AUTO_INCREMENT PRIMARY KEY,
    nombre VARCHAR(60) NOT NULL,
    apellido_paterno VARCHAR(60) NOT NULL,
    apellido_materno VARCHAR(60),
    cedula VARCHAR(30) NOT NULL UNIQUE,
    especialidad VARCHAR(100),
    telefono VARCHAR(20),
    correo VARCHAR(120)
);

CREATE TABLE usuario
(
    id_usuario INT AUTO_INCREMENT PRIMARY KEY,
    id_rol INT NOT NULL,
    id_veterinario INT NULL,
    nombre_completo VARCHAR(120) NOT NULL,
    nombre_usuario VARCHAR(50) NOT NULL UNIQUE,
    contrasena VARCHAR(255) NOT NULL, -- guardar HASH, nunca contraseña en texto plano
    correo VARCHAR(120),
    fecha_registro DATETIME NOT NULL DEFAULT CURRENT_TIMESTAMP,

    CONSTRAINT fk_usuario_rol
        FOREIGN KEY (id_rol)
        REFERENCES rol(id_rol)
        ON UPDATE CASCADE
        ON DELETE RESTRICT,

    CONSTRAINT fk_usuario_veterinario
        FOREIGN KEY (id_veterinario)
        REFERENCES veterinario(id_veterinario)
        ON UPDATE CASCADE
        ON DELETE RESTRICT,

    CONSTRAINT uq_usuario_veterinario
        UNIQUE (id_veterinario)
);

CREATE TABLE mascota
(
    id_mascota INT AUTO_INCREMENT PRIMARY KEY,
    id_cliente INT NOT NULL,
    nombre VARCHAR(60) NOT NULL,
    especie VARCHAR(40) NOT NULL,
    raza VARCHAR(60),
    sexo VARCHAR(15),
    fecha_nacimiento DATE,
    color VARCHAR(50),
    senas_particulares VARCHAR(200),

    CONSTRAINT fk_mascota_cliente
        FOREIGN KEY (id_cliente)
        REFERENCES cliente(id_cliente)
        ON UPDATE CASCADE
        ON DELETE RESTRICT
);

CREATE TABLE consulta
(
    id_consulta INT AUTO_INCREMENT PRIMARY KEY,
    id_mascota INT NOT NULL,
    id_veterinario INT NOT NULL,
    fecha DATETIME NOT NULL DEFAULT CURRENT_TIMESTAMP,
    motivo_consulta VARCHAR(250),
    peso DECIMAL(6,2),
    temperatura DECIMAL(4,1),
    sintomas TEXT,
    diagnostico TEXT,
    tratamiento TEXT,
    observaciones TEXT,

    CONSTRAINT fk_consulta_mascota
        FOREIGN KEY (id_mascota)
        REFERENCES mascota(id_mascota)
        ON UPDATE CASCADE
        ON DELETE RESTRICT,

    CONSTRAINT fk_consulta_veterinario
        FOREIGN KEY (id_veterinario)
        REFERENCES veterinario(id_veterinario)
        ON UPDATE CASCADE
        ON DELETE RESTRICT
);

CREATE TABLE receta
(
    id_receta INT AUTO_INCREMENT PRIMARY KEY,
    id_consulta INT NOT NULL UNIQUE,
    fecha DATETIME NOT NULL DEFAULT CURRENT_TIMESTAMP,
    indicaciones_generales TEXT,

    CONSTRAINT fk_receta_consulta
        FOREIGN KEY (id_consulta)
        REFERENCES consulta(id_consulta)
        ON UPDATE CASCADE
        ON DELETE RESTRICT
);

CREATE TABLE medicamento
(
    id_medicamento INT AUTO_INCREMENT PRIMARY KEY,
    nombre VARCHAR(100) NOT NULL,
    presentacion VARCHAR(100),
    concentracion VARCHAR(60)
);

CREATE TABLE detalle_receta
(
    id_detalle INT AUTO_INCREMENT PRIMARY KEY,
    id_receta INT NOT NULL,
    id_medicamento INT NOT NULL,
    dosis VARCHAR(80) NOT NULL,
    frecuencia VARCHAR(80),
    duracion VARCHAR(80),
    via_administracion VARCHAR(60),
    indicaciones VARCHAR(250),

    CONSTRAINT fk_detalle_receta
        FOREIGN KEY (id_receta)
        REFERENCES receta(id_receta)
        ON UPDATE CASCADE
        ON DELETE RESTRICT,

    CONSTRAINT fk_detalle_medicamento
        FOREIGN KEY (id_medicamento)
        REFERENCES medicamento(id_medicamento)
        ON UPDATE CASCADE
        ON DELETE RESTRICT
);
```


## D-03. Interpretar las relaciones

| **Relación** | **Cardinalidad** | **FK** |
| --- | --- | --- |
| Cliente → Mascota | 1 : N | mascota.id_cliente |
| Mascota → Consulta | 1 : N | consulta.id_mascota |
| Veterinario → Consulta | 1 : N | consulta.id_veterinario |
| Consulta → Receta | 1 : 0..1 | receta.id_consulta UNIQUE |
| Receta → DetalleReceta | 1 : N | detalle_receta.id_receta |
| Medicamento → DetalleReceta | 1 : N | detalle_receta.id_medicamento |
| Rol → Usuario | 1 : N | usuario.id_rol |
| Veterinario → Usuario | 1 : 0..1 | usuario.id_veterinario UNIQUE/NULL |

```sql
USE Veterinaria;

-- PRUEBA 1. Clientes
SELECT *
FROM cliente
ORDER BY apellido_paterno, nombre;

-- PRUEBA 2. Cliente y sus mascotas
SELECT
    c.id_cliente,
    CONCAT(c.nombre, ' ', c.apellido_paterno) AS cliente,
    m.id_mascota,
    m.nombre AS mascota,
    m.especie,
    m.raza
FROM cliente c
INNER JOIN mascota m
    ON c.id_cliente = m.id_cliente
ORDER BY c.id_cliente, m.nombre;

-- PRUEBA 3. Expediente básico
SELECT
    m.nombre AS mascota,
    c.fecha,
    c.peso,
    c.temperatura,
    c.sintomas,
    c.diagnostico,
    c.tratamiento
FROM mascota m
INNER JOIN consulta c
    ON m.id_mascota = c.id_mascota
WHERE m.id_mascota = 1;

-- PRUEBA 4. Receta con medicamentos
SELECT
    m.nombre AS mascota,
    med.nombre AS medicamento,
    dr.dosis,
    dr.frecuencia,
    dr.duracion
FROM mascota m
INNER JOIN consulta c
    ON m.id_mascota = c.id_mascota
INNER JOIN receta r
    ON c.id_consulta = r.id_consulta
INNER JOIN detalle_receta dr
    ON r.id_receta = dr.id_receta
INNER JOIN medicamento med
    ON dr.id_medicamento = med.id_medicamento
WHERE m.id_mascota = 1;

-- PRUEBA 5. Integridad referencial.
-- Ejecutar esta sentencia POR SEPARADO.
-- Debe producir error porque el cliente 999 no existe.
--
-- INSERT INTO mascota
-- (id_cliente, nombre, especie)
-- VALUES
-- (999, 'Prueba', 'Perro');
```

### D. Escribir la cadena

#### Caso 1. `secrets.json` está vacío

Si contiene únicamente:

```json
{}
```

reemplazarlo por:

```json
{
  "ConnectionStrings": {
    "Veterinaria": "Server=localhost;Port=3306;Database=Veterinaria;User ID=TU_USUARIO;Password=TU_PASSWORD;"
  }
}
```

#### Caso 2. `secrets.json` ya contiene otros secretos

**No borrar el contenido existente.**

Agregar la propiedad `ConnectionStrings` dentro del mismo objeto JSON.

Ejemplo:

```json
{
  "OtraConfiguracion": "valor-existente",
  "ConnectionStrings": {
    "Veterinaria": "Server=localhost;Port=3306;Database=Veterinaria;User ID=TU_USUARIO;Password=TU_PASSWORD;"
  }
}
```

> Cuidar las comas y llaves del JSON. Si Visual Studio marca error de formato, revisar que cada propiedad esté separada por coma y que todas las llaves tengan cierre.

Sustituir con los datos reales **de esa computadora**.

Ejemplo:

```text
Hostname = localhost
Port     = 3306
Username = root
```

entonces:

```json
{
  "ConnectionStrings": {
    "Veterinaria": "Server=localhost;Port=3306;Database=Veterinaria;User ID=root;Password=SU_CLAVE_LOCAL;"
  }
}
```

6. Guardar con `Ctrl + S`.
