## F-02. Convertir `Clientes/Index.cshtml.cs` a datos reales

Abrir:

```text
Veterinaria.Web
↓
Pages
↓
Clientes
↓
Index.cshtml.cs
```

En la Práctica 5 existía una lista temporal.

Ahora reemplazar el contenido por:

```csharp
using Microsoft.AspNetCore.Mvc;
using Microsoft.AspNetCore.Mvc.RazorPages;
using Veterinaria.Entidades;
using Veterinaria.Negocios;

namespace Veterinaria.Web.Pages.Clientes;

public class IndexModel : PageModel
{
    [BindProperty(SupportsGet = true)]
    public string? Buscar { get; set; }

    public List<Cliente> Clientes { get; private set; } = new();

    public string? Error { get; private set; }

    public void OnGet()
    {
        try
        {
            ClienteNegocio negocio = new();

            Clientes = string.IsNullOrWhiteSpace(Buscar)
                ? negocio.Listar()
                : negocio.Buscar(Buscar);
        }
        catch (Exception ex)
        {
            System.Diagnostics.Debug.WriteLine(ex);

            Error =
                "No fue posible consultar los clientes. " +
                "Verifique la conexión con MySQL.";
        }
    }
}
```

### Qué cambió

Antes:

## F-11. Convertir `Mascotas/Index.cshtml.cs`

La pantalla debe mostrar:

```text
Mascota
+
Nombre del propietario
```

Se conservará `MascotaFila` como estructura de presentación.

Abrir:

```text
Pages
↓
Mascotas
↓
Index.cshtml.cs
```

Reemplazar por:

```csharp
using Microsoft.AspNetCore.Mvc;
using Microsoft.AspNetCore.Mvc.RazorPages;
using Veterinaria.Entidades;
using Veterinaria.Negocios;

namespace Veterinaria.Web.Pages.Mascotas;

public class IndexModel : PageModel
{
    [BindProperty(SupportsGet = true)]
    public int? IdCliente { get; set; }

    public List<MascotaFila> Mascotas { get; private set; } = new();

    public string? Error { get; private set; }

    public void OnGet()
    {
        try
        {
            MascotaNegocio mascotaNegocio = new();
            ClienteNegocio clienteNegocio = new();

            List<Mascota> mascotas =
                IdCliente.HasValue
                    ? mascotaNegocio.ListarPorCliente(IdCliente.Value)
                    : mascotaNegocio.Listar();

            Dictionary<int, Cliente> clientes =
                clienteNegocio
                    .Listar()
                    // IdCliente es PK; no existen duplicados.
                    .ToDictionary(c => c.IdCliente);

            Mascotas = mascotas
                .Select(m => new MascotaFila
                {
                    Mascota = m,
                    Propietario =
                        clientes.TryGetValue(
                            m.IdCliente,
                            out Cliente? cliente)
                        ? $"{cliente.Nombre} " +
                          $"{cliente.ApellidoPaterno}"
                        : "No disponible"
                })
                .ToList();
        }
        catch (Exception ex)
        {
            System.Diagnostics.Debug.WriteLine(ex);

            Error =
                "No fue posible consultar las mascotas.";
        }
    }

    public class MascotaFila
    {
        public Mascota Mascota { get; set; } = new();

        public string Propietario { get; set; } =
            string.Empty;
    }
}
```

## F-13. Convertir `Mascotas/Nueva.cshtml.cs`

Los propietarios ya no serán:

```text
Ana López
Carlos Ramírez
```

escritos manualmente.

Ahora se obtendrán de MySQL.

Reemplazar por:

```csharp
using Microsoft.AspNetCore.Mvc;
using Microsoft.AspNetCore.Mvc.RazorPages;
using Microsoft.AspNetCore.Mvc.Rendering;
using Veterinaria.Entidades;
using Veterinaria.Negocios;

namespace Veterinaria.Web.Pages.Mascotas;

public class NuevaModel : PageModel
{
    [BindProperty]
    public Mascota Mascota { get; set; } = new();

    public List<SelectListItem> Clientes { get; private set; } = new();

    public bool HayClientes => Clientes.Count > 0;

    public string? Error { get; private set; }

    public void OnGet()
    {
        try
        {
            CargarClientes();
        }
        catch (Exception ex)
        {
            System.Diagnostics.Debug.WriteLine(ex);

            Error =
                "No fue posible cargar los clientes. " +
                "Verifique la conexión con MySQL.";
        }
    }

    public IActionResult OnPost()
    {
        try
        {
            CargarClientes();
        }
        catch (Exception ex)
        {
            System.Diagnostics.Debug.WriteLine(ex);

            ModelState.AddModelError(
                string.Empty,
                "No fue posible cargar los clientes.");

            return Page();
        }

        if (!HayClientes)
        {
            ModelState.AddModelError(
                string.Empty,
                "No existen clientes registrados. " +
                "Primero registre un cliente.");

            return Page();
        }

        if (!ModelState.IsValid)
            return Page();

        try
        {
            MascotaNegocio negocio = new();

            int id = negocio.Agregar(Mascota);

            TempData["Mensaje"] =
                $"Mascota registrada correctamente. ID: {id}";

            return RedirectToPage("./Index");
        }
        catch (ArgumentException ex)
        {
            ModelState.AddModelError(
                string.Empty,
                ex.Message);

            return Page();
        }
        catch (Exception ex)
        {
            System.Diagnostics.Debug.WriteLine(ex);

            ModelState.AddModelError(
                string.Empty,
                "No fue posible registrar la mascota.");

            return Page();
        }
    }

    private void CargarClientes()
    {
        ClienteNegocio negocio = new();

        Clientes = negocio
            .Listar()
            .Select(c => new SelectListItem
            {
                Value = c.IdCliente.ToString(),
                Text =
                    $"{c.Nombre} {c.ApellidoPaterno} {c.ApellidoMaterno}".Trim()
            })
            .ToList();
    }
}
```

El botón **Guardar** puede deshabilitarse cuando no existan clientes:

```html
<button type="submit"
        class="btn btn-primary"
        disabled="@(!Model.HayClientes)">
    Guardar
</button>
```

>
> Cuando la expresión booleana es `true`, Razor agrega el atributo `disabled`.
> Cuando es `false`, lo omite. Por eso se utiliza `!Model.HayClientes`.

```csharp
using Microsoft.AspNetCore.Mvc;
using Microsoft.AspNetCore.Mvc.RazorPages;
using Veterinaria.Entidades;
using Veterinaria.Negocios;

namespace Veterinaria.Web.Pages.Clientes;

public class EliminarModel : PageModel
{
    public Cliente Cliente { get; private set; } = new();

    public string? Error { get; private set; }

    public IActionResult OnGet(int id)
    {
        try
        {
            ClienteNegocio negocio = new();

            Cliente? cliente =
                negocio.BuscarPorId(id);

            if (cliente is null)
                return NotFound();

            Cliente = cliente;

            return Page();
        }
        catch (Exception ex)
        {
            System.Diagnostics.Debug.WriteLine(ex);

            Error =
                "No fue posible consultar el cliente.";

            return Page();
        }
    }

    public IActionResult OnPost(int id)
    {
        ClienteNegocio negocio = new();

        Cliente? cliente =
            negocio.BuscarPorId(id);

        if (cliente is null)
            return NotFound();

        Cliente = cliente;

        try
        {
            bool eliminado =
                negocio.Eliminar(id);

            if (!eliminado)
                return NotFound();

            TempData["Mensaje"] =
                "Cliente eliminado correctamente.";

            return RedirectToPage("./Index");
        }
        catch (ReglaNegocioException ex)
        {
            Error = ex.Message;

            return Page();
        }
        catch (Exception ex)
        {
            System.Diagnostics.Debug.WriteLine(ex);

            Error =
                "No fue posible eliminar el cliente.";

            return Page();
        }
    }
}
```

## I-11. Programar `Mascotas/Eliminar.cshtml.cs`

```csharp
using Microsoft.AspNetCore.Mvc;
using Microsoft.AspNetCore.Mvc.RazorPages;
using Veterinaria.Entidades;
using Veterinaria.Negocios;

namespace Veterinaria.Web.Pages.Mascotas;

public class EliminarModel : PageModel
{
    public Mascota Mascota { get; private set; } = new();

    public Cliente Propietario { get; private set; } = new();

    public string? Error { get; private set; }

    public IActionResult OnGet(int id)
    {
        try
        {
            return Cargar(id);
        }
        catch (Exception ex)
        {
            System.Diagnostics.Debug.WriteLine(ex);

            Error =
                "No fue posible consultar la mascota.";

            return Page();
        }
    }

    public IActionResult OnPost(int id)
    {
        IActionResult resultado =
            Cargar(id);

        if (resultado is NotFoundResult)
            return resultado;

        try
        {
            MascotaNegocio negocio = new();

            bool eliminada =
                negocio.Eliminar(id);

            if (!eliminada)
                return NotFound();

            TempData["Mensaje"] =
                "Mascota eliminada correctamente.";

            return RedirectToPage("./Index");
        }
        catch (ReglaNegocioException ex)
        {
            Error = ex.Message;

            return Page();
        }
        catch (Exception ex)
        {
            System.Diagnostics.Debug.WriteLine(ex);

            Error =
                "No fue posible eliminar la mascota.";

            return Page();
        }
    }

    private IActionResult Cargar(int id)
    {
        MascotaNegocio mascotaNegocio = new();

        Mascota? mascota =
            mascotaNegocio.BuscarPorId(id);

        if (mascota is null)
            return NotFound();

        Mascota = mascota;

        ClienteNegocio clienteNegocio = new();

        Cliente? propietario =
            clienteNegocio.BuscarPorId(
                mascota.IdCliente);

        if (propietario is null)
            return NotFound();

        Propietario = propietario;

        return Page();
    }
}
```
