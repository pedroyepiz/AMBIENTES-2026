## B-05. Revisar manejo de errores

La aplicación debe utilizar un criterio consistente según el tipo de página:

```text
FORMULARIO EDITABLE
Nuevo / Editar
→ Data Annotations / ModelState
→ conservar los datos capturados

PÁGINA DE CONSULTA
Index / Detalle / carga inicial
→ propiedad Error
→ mensaje visible

PÁGINA ELIMINAR
ReglaNegocioException
→ propiedad Error

ERROR TÉCNICO
→ catch (Exception ex)
→ Debug.WriteLine(ex)
→ mensaje amigable
→ nunca mostrar el detalle técnico al usuario
```

> `ModelState` y `Error` pueden existir en el mismo proyecto. La regla es **no utilizarlos de forma contradictoria dentro de la misma operación**.

No mostrar al usuario:

```text
ex.ToString()
stack trace
password
connection string
mensaje técnico completo de MySQL
```

### Patrón homologado para páginas `Eliminar`

Las páginas de eliminación de P6 ya utilizan la propiedad:

```csharp
public string? Error { get; private set; }
```

Ese patrón debe conservarse.

El `PageModel` debe mantener la propiedad:

```csharp
public string? Error { get; private set; }
```

y utilizarla tanto para reglas de negocio como para mensajes técnicos amigables.

> No es necesario agregar `using System.Diagnostics;` porque se utiliza el nombre completo:
>
> ```csharp
> System.Diagnostics.Debug.WriteLine(ex);
> ```

> **Importante al regresar `Page()`**
>
> Antes de intentar la eliminación, el `PageModel` debe cargar los datos necesarios para volver a dibujar la pantalla si ocurre un error, tal como quedó implementado en P6.
>
> Por ejemplo, una página de Mascota debe conservar disponibles la Mascota y su Propietario antes de ejecutar `Eliminar(id)`.

### Corrección importante respecto a P6

En P6, las páginas de eliminación cargaban parte de la información **antes** del `try` del `OnPost`.

En P7 esa carga debe quedar dentro del mismo bloque protegido, para que una falla de MySQL no termine en una página 500.

#### `Clientes/Eliminar.cshtml.cs` — `OnPost`

```csharp
public IActionResult OnPost(int id)
{
    try
    {
        ClienteNegocio negocio = new();

        Cliente? cliente =
            negocio.BuscarPorId(id);

        if (cliente is null)
            return NotFound();

        Cliente = cliente;

        bool eliminado =
            negocio.Eliminar(id);

        if (!eliminado)
            return NotFound();

        TempData["Mensaje"] =
            "Cliente eliminado correctamente.";

        return RedirectToPage("./Index");
    }
    catch (ReglaNegocioException ex)
    {
        Error = ex.Message;
        return Page();
    }
    catch (Exception ex)
    {
        System.Diagnostics.Debug.WriteLine(ex);

        Error =
            "No fue posible eliminar el cliente.";

        return Page();
    }
}
```

#### `Mascotas/Eliminar.cshtml.cs` — `OnPost`

```csharp
public IActionResult OnPost(int id)
{
    try
    {
        IActionResult resultado =
            Cargar(id);

        if (resultado is NotFoundResult)
            return resultado;

        MascotaNegocio negocio = new();

        bool eliminada =
            negocio.Eliminar(id);

        if (!eliminada)
            return NotFound();

        TempData["Mensaje"] =
            "Mascota eliminada correctamente.";

        return RedirectToPage("./Index");
    }
    catch (ReglaNegocioException ex)
    {
        Error = ex.Message;
        return Page();
    }
    catch (Exception ex)
    {
        System.Diagnostics.Debug.WriteLine(ex);

        Error =
            "No fue posible eliminar la mascota.";

        return Page();
    }
}
```

> Esta corrección **no cambia la arquitectura ni el alcance**. Sólo hace consistente el manejo de errores que P7 está auditando.

La vista debe utilizar el **mismo canal de error**:

```html
@if (!string.IsNullOrWhiteSpace(Model.Error))
{
    <div class="alert alert-danger">
        @Model.Error
    </div>

    <a asp-page="./Index"
       class="btn btn-secondary">
        Regresar
    </a>
}
else
{
    <div class="alert alert-warning">
        Confirmar eliminación
    </div>

    <form method="post">
        <button type="submit"
                class="btn btn-danger">
            Eliminar
        </button>

        <a asp-page="./Index"
           class="btn btn-secondary">
            Cancelar
        </a>
    </form>
}
```

### Regla de consistencia

```text
PageModel usa Error
        ↓
Vista lee Model.Error
        ↓
mensaje visible
        ↓
botón Eliminar oculto cuando existe error
```

No mezclar en estas páginas:

```text
ModelState.AddModelError(...)
```

con:

```text
Model.Error
```

para representar la **misma regla de eliminación**, porque la vista podría no mostrar el mensaje esperado.

> Las validaciones de formularios, por ejemplo correo inválido o campos obligatorios, sí continúan utilizando `ModelState` y Data Annotations.

```sql
USE Veterinaria;

-- =========================================================
-- 1. TOTAL DE TABLAS
-- Esperado: 9
-- =========================================================

SELECT COUNT(*) AS total_tablas
FROM information_schema.tables
WHERE table_schema = 'Veterinaria'
  AND table_type = 'BASE TABLE';


-- =========================================================
-- 2. NOMBRES DE LAS 9 TABLAS
-- =========================================================

SELECT table_name
FROM information_schema.tables
WHERE table_schema = 'Veterinaria'
  AND table_type = 'BASE TABLE'
ORDER BY table_name;


-- =========================================================
-- 3. CLAVES PRIMARIAS
-- Esperado: 9 PK
-- =========================================================

SELECT
    table_name,
    column_name
FROM information_schema.key_column_usage
WHERE table_schema = 'Veterinaria'
  AND constraint_name = 'PRIMARY'
ORDER BY table_name;


-- =========================================================
-- 4. LLAVES FORÁNEAS
-- Esperado: 8 FK
-- =========================================================

SELECT
    table_name,
    column_name,
    constraint_name,
    referenced_table_name,
    referenced_column_name
FROM information_schema.key_column_usage
WHERE table_schema = 'Veterinaria'
  AND referenced_table_name IS NOT NULL
ORDER BY table_name, column_name;


-- =========================================================
-- 5. REGLAS DE INTEGRIDAD REFERENCIAL
-- Esperado en P5:
-- UPDATE = CASCADE
-- DELETE = RESTRICT
-- =========================================================

SELECT
    table_name,
    constraint_name,
    referenced_table_name,
    update_rule,
    delete_rule
FROM information_schema.referential_constraints
WHERE constraint_schema = 'Veterinaria'
ORDER BY table_name, constraint_name;


-- =========================================================
-- 6. CONTEO INFORMATIVO DE REGISTROS
-- Un valor 0 NO significa automáticamente que exista un error.
-- Por ejemplo, usuario puede estar vacío en los datos de prueba.
-- =========================================================

SELECT 'rol' AS tabla, COUNT(*) AS filas
FROM rol

UNION ALL
SELECT 'usuario', COUNT(*)
FROM usuario

UNION ALL
SELECT 'cliente', COUNT(*)
FROM cliente

UNION ALL
SELECT 'mascota', COUNT(*)
FROM mascota

UNION ALL
SELECT 'veterinario', COUNT(*)
FROM veterinario

UNION ALL
SELECT 'consulta', COUNT(*)
FROM consulta

UNION ALL
SELECT 'receta', COUNT(*)
FROM receta

UNION ALL
SELECT 'medicamento', COUNT(*)
FROM medicamento

UNION ALL
SELECT 'detalle_receta', COUNT(*)
FROM detalle_receta;


-- =========================================================
-- 7. CLIENTES
-- =========================================================

SELECT
    id_cliente,
    nombre,
    apellido_paterno,
    telefono,
    correo
FROM cliente
ORDER BY id_cliente;


-- =========================================================
-- 8. MASCOTAS CON PROPIETARIO
-- =========================================================

SELECT
    m.id_mascota,
    m.nombre AS mascota,
    m.especie,
    m.raza,
    CONCAT(
        c.nombre,
        ' ',
        c.apellido_paterno
    ) AS propietario
FROM mascota m
INNER JOIN cliente c
    ON c.id_cliente = m.id_cliente
ORDER BY m.id_mascota;


-- =========================================================
-- 9. MASCOTAS CON CONSULTAS
-- Max debe conservar al menos su consulta de datos iniciales.
-- =========================================================

SELECT
    m.id_mascota,
    m.nombre AS mascota,
    COUNT(q.id_consulta) AS consultas
FROM mascota m
LEFT JOIN consulta q
    ON q.id_mascota = m.id_mascota
GROUP BY
    m.id_mascota,
    m.nombre
ORDER BY m.id_mascota;


-- =========================================================
-- 10. DATOS TEMPORALES DE ACEPTACIÓN
-- Al finalizar A-13 y A-14, ambas consultas deben devolver 0 filas.
-- =========================================================

SELECT *
FROM cliente
WHERE correo = 'prueba.p7@local.test';

SELECT *
FROM mascota
WHERE nombre = 'PixelP7';
```

La sección 6 sirve para comprobar que las tablas pueden consultarse y para observar su estado.

No se debe afirmar:

porque una tabla puede existir correctamente y encontrarse vacía.

En particular:

```text
usuario
```

```text
0 filas
```

en los datos académicos de prueba y eso **no invalida la estructura**.

Lo que sí debe coincidir obligatoriamente es:

```text
✓ 9 tablas con los nombres esperados
✓ 9 claves primarias
✓ 8 llaves foráneas
✓ relaciones correctas
✓ reglas referenciales correctas
```

# 13. ⚠️ Errores frecuentes

| Error | Corrección |
| --- | --- |
| P7 empieza a agregar nuevos módulos | Detenerse; P7 es pruebas y cierre |
| `main` tiene errores | Corregir antes de aceptación |
| Frontend rediseña todo | Corregir sólo defectos necesarios |
| El propietario muestra `Veterinaria.Entidades.Cliente` | Mostrar propiedades del Cliente |
| Fecha futura se guarda | Revisar `MascotaNegocio.Validar()` |
| Cliente con Mascota se elimina | Revisar `TieneMascotas()` y `ClienteNegocio.Eliminar()` |
| Max se elimina | Revisar `TieneConsultas()` y `MascotaNegocio.Eliminar()` |
| Error técnico aparece al usuario | Registrar con `Debug.WriteLine(ex)`, ejecutar con **F5** para verlo en Output y mostrar al usuario sólo un mensaje amigable |
| `ReglaNegocioException` aparece en rojo | Verificar que exista `Veterinaria.Negocios/ReglaNegocioException.cs` y `using Veterinaria.Negocios;` |
| El `catch (ReglaNegocioException)` nunca se ejecuta | Confirmar que `ClienteNegocio.Eliminar()` y `MascotaNegocio.Eliminar()` lancen `ReglaNegocioException` y no `InvalidOperationException` |
| SQL aparece en Razor Pages | Mover a Datos |
| `MySqlConnection` aparece en Web | Mover acceso a Datos |
| Workbench no muestra el cambio | Confirmar que Web y Workbench usan la misma base local |
| Aparecen varios `PRUEBA P7` | Reconstruir la base antes de repetir aceptación |
| El backup está vacío | Repetir Data Export |
| El conteo dice 9 tablas, pero falta una tabla esperada | Comparar también los nombres; cantidad correcta no garantiza estructura correcta |
| `usuario` muestra 0 filas | No asumir error: verificar que la tabla exista, tenga PK/FK correctas y recordar que los datos de prueba pueden no insertar usuarios |
| Faltan PK o FK en `04_verificacion_final_p7.sql` | Comparar el resultado con 9 PK y 8 FK del modelo definido en P5 |
| Contraseña aparece en README | Eliminarla y usar User Secrets |
| Otra computadora no ve cambios Git | Fetch + Pull |
| Rama está atrasada | Update from main |
| `main` no tiene un trabajo | Falta Pull Request / Merge |
| Se borró una rama | No era necesario borrar ramas |
| Matriz tiene celdas vacías | Completar antes de entregar |
| `Debug.WriteLine` no aparece | Confirmar configuración **Debug**, ejecutar con **F5** y revisar `Output → Debug` |
| `OnGet` captura un error pero la vista queda vacía | Mostrar `Model.Error` y ocultar el contenido normal cuando corresponda |
| Una falla de MySQL en `Eliminar` termina en página 500 | Verificar que la carga de Cliente/Mascota esté dentro del `try` del `OnPost` |
| El respaldo final no coincide con la aceptación | Generarlo en la computadora oficial sobre la misma base aprobada |
| Dos integrantes editan el mismo documento | Aplicar “un archivo, un responsable, su rama” |
| `.vs/`, `bin/` u `obj/` aparecen en Changes | Corregir `.gitignore` antes de hacer Commit |
| Un caso de aceptación falla y se corrige en `backend` | Detenerse y corregir en la rama del rol responsable; integrar por Pull Request |
| Evidencias vienen de bases distintas | Utilizar la computadora oficial de aceptación |
