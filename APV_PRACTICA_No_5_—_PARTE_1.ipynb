Esta práctica está escrita para alumnos que todavía están aprendiendo a utilizar Git, GitHub y GitHub Desktop.

# Ruta rápida de la práctica

```text
1. Preparar las 3 cuentas
        ↓
2. Backend crea el repositorio Veterinaria
        ↓
3. Backend publica el repositorio
        ↓
4. Backend agrega Frontend y DBA
        ↓
5. Backend crea las ramas
   backend / frontend / dba
        ↓
6. Frontend y DBA clonan
        ↓
7. Cada integrante trabaja en su rama
        ↓
8. Cada integrante hace Commit + Push
        ↓
9. Se comprueba el aislamiento de las ramas
        ↓
10. Se crean Pull Requests
        ↓
11. Se revisan e integran a main
        ↓
12. Los tres actualizan main
        ↓
13. Los tres actualizan su rama de trabajo
        ↓
14. Prueba final de sincronización
```

# 9. Conceptos que conviene conocer antes de empezar

| Concepto | Significado en esta práctica |

<a id="38-diferencia-entre-commit-y-push"></a>
# 38. Diferencia entre Commit y Push

Antes de Push:

```text
ARCHIVO
   ↓
COMMIT
   ↓
REPOSITORIO LOCAL
```

Después de Push:

```text
COMPUTADORA
    ↓
   PUSH
    ↓
  GITHUB
```

Por lo tanto:

```text
COMMIT ≠ PUSH
```

Un Commit guarda.

Un Push publica ese commit en el repositorio remoto.

<a id="59-errores-frecuentes-y-como-corregirlos"></a>
# 59. Errores frecuentes y cómo corregirlos

| Problema | Qué revisar |
| --- | --- |
| No aparece el repositorio para clonar | Confirmar que el alumno aceptó la invitación y está en la cuenta correcta |
| `main` no aparece como rama inicial | Revisar `File → Options → Git → Default branch` antes de crear el repositorio |
| No aparece `.gitignore` | Revisar si se seleccionó `VisualStudio` al crear el repositorio |
| El alumno modificó archivos estando en `main` | Detenerse antes de Commit y avisar al profesor |
| No aparece `Push origin` | Puede que no existan commits nuevos; revisar History |
| El archivo está en la rama equivocada | Revisar Current Branch y no integrar hasta corregir |
| Frontend no ve el commit del Backend | Las ramas son independientes; cambiar a la rama correcta o esperar la integración |
| `main` todavía no contiene los archivos | Revisar que los Pull Requests hayan sido integrados |
| Una computadora no tiene los tres archivos | Cambiar a `main`, Fetch y Pull |
| La rama de trabajo quedó desactualizada | Utilizar `Branch → Update from default branch` |
| GitHub ofrece `Delete branch` | No eliminar las ramas en esta práctica |
| El Pull Request apunta a otra rama | Corregir Base=`main` y Compare=`rama del alumno` |
| El autor intenta revisar su propio PR | Otro integrante debe revisar ese Pull Request |
| Hay cambios pendientes al cambiar de rama | Guardar/commit o consultar al profesor antes de mover cambios entre ramas |
