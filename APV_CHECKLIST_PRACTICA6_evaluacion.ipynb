# 🐍 Evaluación automática en Python

Ejecuta la siguiente celda **después de revisar los 10 puntos**.

El programa solicitará:

```text
Equipo
Grupo
Nombre de Frontend
Nombre de Backend / Líder
Nombre de DBA / Pruebas
```

Luego preguntará:

```text
Puntos 1–9:
1 = Cumplido
0 = No cumplido

Punto 10:
4 aspectos separados
cada uno vale 0.25
```

Al final mostrará la calificación final sobre 10 y un resumen listo para copiar.

[⬆️ Volver al menú](#scrollTo=p6_eval_menu)

````python
def pedir_texto(mensaje):
    while True:
        valor = input(mensaje).strip()
        if valor:
            return valor
        print("⚠️ El dato no puede quedar vacío.")

def pedir_binario(mensaje):
    while True:
        valor = input(mensaje + " [1 = Sí / 0 = No]: ").strip()
        if valor in ("0", "1"):
            return int(valor)
        print("⚠️ Escriba solamente 1 o 0.")

print("=" * 70)
print("📘 APV — PRÁCTICA 6")
print("✅ CHECKLIST AUTOMÁTICO DE EVALUACIÓN")
print("=" * 70)

equipo = pedir_texto("Nombre o número del equipo: ")
grupo = pedir_texto("Grupo: ")
frontend = pedir_texto("Nombre del integrante FRONTEND: ")
backend = pedir_texto("Nombre del integrante BACKEND / LÍDER: ")
dba = pedir_texto("Nombre del integrante DBA / PRUEBAS: ")

criterios = [
    "Punto 1 — main, compilación y arranque",
    "Punto 2 — Workbench y las 9 tablas",
    "Punto 3 — Clientes",
    "Punto 4 — Mascotas",
    "Punto 5 — Fecha futura",
    "Punto 6 — Cliente con Mascota NO se elimina",
    "Punto 7 — Max NO se elimina",
    "Punto 8 — Eliminación válida y limpieza",
    "Punto 9 — Arquitectura + GitHub",
]

print("\n" + "=" * 70)
print("🧪 PUNTOS 1 AL 9")
print("=" * 70)

resultados = [pedir_binario(c) for c in criterios]

print("\n" + "=" * 70)
print("🎨 PUNTO 10 — FRONTEND")
print("=" * 70)

subcriterios = [
    "Diseño visual",
    "Usabilidad",
    "Responsive",
    "Creatividad / presentación",
]
res_frontend = [pedir_binario(c) for c in subcriterios]

puntos_1_9 = sum(resultados)
punto_10 = sum(res_frontend) * 0.25
calificacion = puntos_1_9 + punto_10

if calificacion >= 9:
    estado = "🏆 EXCELENTE"
elif calificacion >= 8:
    estado = "✅ MUY BIEN"
elif calificacion >= 7:
    estado = "✅ APROBADO"
elif calificacion >= 6:
    estado = "⚠️ APROBADO CON ÁREAS DE MEJORA"
else:
    estado = "❌ PENDIENTE / NO APROBADO"

print("\n\n" + "=" * 70)
print("📋 RESULTADO DE LA EVALUACIÓN")
print("=" * 70)
print(f"Equipo: {equipo}")
print(f"Grupo: {grupo}")
print(f"🖥️ Frontend: {frontend}")
print(f"⚙️ Backend / Líder: {backend}")
print(f"🗄️ DBA / Pruebas: {dba}")

print("\nRESULTADOS:")
for i, (criterio, valor) in enumerate(zip(criterios, resultados), start=1):
    marca = "✅ CUMPLIDO" if valor else "❌ NO CUMPLIDO"
    print(f"{i:>2}. {marca} | {criterio}")

print("\n10. 🎨 FRONTEND")
for criterio, valor in zip(subcriterios, res_frontend):
    marca = "✅" if valor else "❌"
    pts = 0.25 if valor else 0.0
    print(f"    {marca} {criterio:<28} {pts:.2f}")

print("\n" + "-" * 70)
print(f"Puntos 1–9:        {puntos_1_9:.2f} / 9.00")
print(f"Punto 10 Frontend: {punto_10:.2f} / 1.00")
print(f"CALIFICACIÓN FINAL: {calificacion:.2f} / 10.00")
print(f"ESTADO: {estado}")
print("-" * 70)

print("\n📄 RESUMEN PARA REPORTE")
print(f"""
Equipo: {equipo}
Grupo: {grupo}

Frontend: {frontend}
Backend / Líder: {backend}
DBA / Pruebas: {dba}

Puntos funcionales: {puntos_1_9:.2f} / 9.00
Frontend — diseño y creatividad: {punto_10:.2f} / 1.00
Calificación Práctica 6: {calificacion:.2f} / 10.00
Estado: {estado}
""")



























````
