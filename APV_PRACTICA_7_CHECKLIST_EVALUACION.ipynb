# 🎨 Punto 10 — Frontend original, no genérico
**Valor total: 1.0 · 🖥️ Frontend**

> No se evalúa sólo que Bootstrap funcione. Se evalúa que la interfaz tenga **identidad propia** y no parezca una plantilla genérica sin trabajo de diseño.

| Aspecto | Qué revisar | Valor |
|---|---|---:|
| ✨ Originalidad / identidad | Hay decisiones visuales propias; no es Bootstrap por defecto. | 0.25 |
| 🎨 Consistencia | Inicio, Clientes, Mascotas y formularios comparten una misma línea visual. | 0.25 |
| 🧭 Usabilidad | Menú, botones y acciones son claros y fáciles de localizar. | 0.25 |
| 📱 Responsive y acabado | Funciona en móvil/escritorio sin elementos rotos o descuidados. | 0.25 |

```text

```

### ❌ No cumple originalidad si
- sólo usa Bootstrap por defecto;

Compara Inicio, Clientes, Mascotas, Nuevo, Editar, Detalle y Eliminar.

# 🐍 Calificación automática

Ejecuta la siguiente celda después de terminar la revisión.

Solicita:
- equipo y grupo;
- nombre de Frontend;
- nombre de Backend/Líder;
- nombre de DBA/Pruebas;
- puntos 1–9 con `1` o `0`;
- los 4 aspectos del Punto 10.

Al final genera la calificación sobre 10 y un resumen.

[⬆️ Volver al menú](#scrollTo=p7_eval_menu)

````python
def texto(msg):
    while True:
        v = input(msg).strip()
        if v:
            return v
        print("⚠️ No deje el dato vacío.")

def si_no(msg):
    while True:
        v = input(msg + " [1 = Sí / 0 = No]: ").strip()
        if v in ("0","1"):
            return int(v)
        print("⚠️ Escriba 1 o 0.")

print("="*76)
print("📘 APV — PRÁCTICA 7 | EVALUACIÓN FINAL")
print("="*76)

equipo = texto("Equipo: ")
grupo = texto("Grupo: ")
frontend = texto("Frontend: ")
backend = texto("Backend / Líder: ")
dba = texto("DBA / Pruebas: ")

criterios = [
    "main final, compilación y ejecución",
    "trabajo colaborativo real",
    "base completa, integridad y respaldo",
    "Clientes final",
    "Mascotas final",
    "reglas de negocio e integridad",
    "arquitectura y manejo de errores",
    "integración final y 21/21 aceptación",
    "documentación y entrega final",
]

print("\n🧪 PUNTOS 1 AL 9")
resultados = [si_no(f"Punto {i} — {c}") for i,c in enumerate(criterios,1)]

sub = [
    "Originalidad / identidad",
    "Consistencia visual",
    "Usabilidad",
    "Responsive y acabado",
]

print("\n🎨 PUNTO 10 — FRONTEND ORIGINAL")
res_front = [si_no(x) for x in sub]

puntos_1_9 = sum(resultados)
punto10 = sum(res_front) * 0.25
cal = puntos_1_9 + punto10

if cal >= 9:
    estado = "🏆 EXCELENTE"
elif cal >= 8:
    estado = "✅ MUY BIEN"
elif cal >= 7:
    estado = "✅ APROBADO"
elif cal >= 6:
    estado = "⚠️ APROBADO CON ÁREAS DE MEJORA"
else:
    estado = "❌ PENDIENTE / NO APROBADO"

print("\n" + "="*76)
print("📋 RESULTADO")
print("="*76)
print(f"Equipo: {equipo} | Grupo: {grupo}")
print(f"🖥️ Frontend: {frontend}")
print(f"⚙️ Backend / Líder: {backend}")
print(f"🗄️ DBA / Pruebas: {dba}")

print("\nPUNTOS 1–9")
for i,(c,v) in enumerate(zip(criterios,resultados),1):
    print(f"{i:>2}. {'✅ CUMPLIDO' if v else '❌ NO CUMPLIDO'} | {c}")

print("\n10. 🎨 FRONTEND")
for c,v in zip(sub,res_front):
    print(f"    {'✅' if v else '❌'} {c:<28} {(0.25 if v else 0):.2f}")

print("\n" + "-"*76)
print(f"Puntos 1–9:              {puntos_1_9:.2f} / 9.00")
print(f"Frontend original/diseño:{punto10:>6.2f} / 1.00")
print(f"CALIFICACIÓN FINAL:       {cal:.2f} / 10.00")
print(f"ESTADO: {estado}")
print("-"*76)

if resultados[1] == 0:
    print("⚠️ No quedó demostrada claramente la colaboración de los 3 integrantes.")
if resultados[7] == 0:
    print("⚠️ No quedó demostrada la integración final / aceptación 21 de 21.")
if punto10 < 0.75:
    print("⚠️ El Frontend requiere mayor originalidad, identidad o acabado.")

print("AMBIENTES DE PROGRAMACION VISUAL")
print("Practica 7 ")
print("🔗 cierre del Sistema Veterinario")
print("\n📄 RESUMEN PARA REPORTE")
print(f"""
Equipo: {equipo}
Grupo: {grupo}
Frontend: {frontend}
Backend / Líder: {backend}
DBA / Pruebas: {dba}

Puntos generales: {puntos_1_9:.2f} / 9.00
Frontend — originalidad y diseño: {punto10:.2f} / 1.00
Calificación Práctica 7: {cal:.2f} / 10.00
Estado: {estado}
""")
````
